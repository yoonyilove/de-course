## Q3. `def · return` — 분류 규칙에 이름을 붙여 다시 쓰기

- `def`는 여러 줄짜리 규칙에 이름을 붙입니다.
- `return`은 함수 안에서 만든 결과를 함수 밖으로 돌려줍니다.
- `print`는 화면에 보여주지만, 그 값을 다음 계산에 쓰게 돌려주는 것은 아닙니다.

### 핵심 빈칸

아래 함수에서 `FILL_ME`를 채워 합계와 등급을 돌려주세요.

```python
def activity_level(login_counts, high_boundary=10, medium_boundary=5):
    total = sum(int(x) for x in login_counts)

    if total >= high_boundary:
        level = "활발"
    elif total >= medium_boundary:
        level = "보통"
    else:
        level = "낮음"

    FILL_ME total, level

total, level = activity_level(members[0]["weekly_logins_raw"])
print(total, level)
```

### `return`을 뺐을 때 복구하기

아래 함수는 화면에는 `보통`을 보여주지만, 바깥 변수 `result`에는 `None`이 들어갑니다. `print(level)`을 고쳐 바깥에서도 `보통`을 사용할 수 있게 하세요.

```python
def broken_activity_level(total):
    if total >= 10:
        level = "활발"
    elif total >= 5:
        level = "보통"
    else:
        level = "낮음"
    print(level)

result = broken_activity_level(8)
print("함수 밖의 결과:", result)
```

```python
predicted_rows = FILL_ME
predicted_active_name = "여기에 이름"

reports = []
for member in FILL_ME:
    total, level = FILL_ME(member["weekly_logins_raw"])
    reports.append({
        "member_id": member["member_id"],
        "name": member["name"],
        "login_total": total,
        "activity_level": level,
    })

print("예상 행 수:", predicted_rows, "실제 행 수:", len(reports))
for report in reports:
    print(report)
```
