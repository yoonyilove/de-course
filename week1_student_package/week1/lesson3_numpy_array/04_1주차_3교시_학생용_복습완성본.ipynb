## Q1. list를 NumPy 2차원 배열로 바꾸면 무엇이 달라질까요?

> **처음 배우는 사람을 위한 안내**
>
> 뜻: 같은 모양의 숫자를 행과 열로 묶습니다.
> 
> 이유: 회원마다 반복문을 복사하지 않고 전체에 같은 계산을 적용하기 위해서입니다.
> 
> 정상: (4, 4), 2, int64 계열이 출력됩니다.
> 
> 복구: 네 행의 길이가 모두 4인지, np.array의 대괄호 짝이 맞는지 확인합니다.

### 실행 전에 생각할 질문

실행 전 예측: 행은 회원, 열은 주차입니다. shape·ndim·dtype을 먼저 적으세요.

### 학생이 할 일

1. 결과를 먼저 적습니다.
2. 핵심 빈칸만 채웁니다.
3. 실행 결과를 예상과 비교합니다.
4. 숫자·조건·열 하나를 바꿔 다시 실행합니다.
5. 결과가 달라진 이유를 한 문장으로 설명합니다.


```python
member_names = ["김철수", "이영희", "박민수", "최하늘"]
login_lists = [[2,3,2,1], [1,0,1,0], [1,2,4,3], [1,0,0,1]]

prediction_shape = "(4, 4)"
login_matrix = np.array(login_lists)
print("예상 shape:", prediction_shape)
print("실제 shape:", login_matrix.shape)
print("ndim:", login_matrix.ndim, "dtype:", login_matrix.dtype)
assert login_matrix.shape == (4, 4)
```

## Q2. 한 회원과 한 주를 정확히 선택하세요

> **처음 배우는 사람을 위한 안내**
>
> 뜻: arr[행, 열]은 한 칸, 콜론(:)은 그 축의 전부를 뜻합니다.
> 
> 이유: 전체 배열에서 업무 질문에 필요한 회원 또는 주차만 골라야 하기 때문입니다.
> 
> 정상: 박민수 4주 값 [1 2 4 3], 3주차 [2 1 4 0]이 출력됩니다.
> 
> 복구: Python 위치는 0부터 시작하고 arr[행, 열] 순서임을 확인합니다.

### 실행 전에 생각할 질문

박민수는 세 번째 사람이므로 행 번호는 2입니다. 3주차 열 번호도 2입니다.

### 학생이 할 일

1. 결과를 먼저 적습니다.
2. 핵심 빈칸만 채웁니다.
3. 실행 결과를 예상과 비교합니다.
4. 숫자·조건·열 하나를 바꿔 다시 실행합니다.
5. 결과가 달라진 이유를 한 문장으로 설명합니다.


```python
minsu_four_weeks = login_matrix[2, :]
week3_all_members = login_matrix[:, 2]
print("박민수:", minsu_four_weeks)
print("3주차:", week3_all_members)
assert minsu_four_weeks.tolist() == [1,2,4,3]
```

## Q4. axis=0과 axis=1은 어떤 질문에 답할까요?

> **처음 배우는 사람을 위한 안내**
>
> 뜻: axis=0은 행 방향으로 내려가 열별 결과, axis=1은 열 방향으로 가며 행별 결과를 냅니다.
> 
> 이유: 주별 평균과 회원별 평균은 같은 mean이라도 축이 다르기 때문입니다.
> 
> 정상: 주별 평균 4개, 회원별 평균 4개가 나옵니다.
> 
> 복구: 결과 개수가 ‘주 수’인지 ‘회원 수’인지 보고 축을 거꾸로 썼는지 확인합니다.

### 실행 전에 생각할 질문

둘 다 길이가 4라 헷갈립니다. 첫 주 평균과 김철수 평균을 손으로 계산해 확인하세요.

### 학생이 할 일

1. 결과를 먼저 적습니다.
2. 핵심 빈칸만 채웁니다.
3. 실행 결과를 예상과 비교합니다.
4. 숫자·조건·열 하나를 바꿔 다시 실행합니다.
5. 결과가 달라진 이유를 한 문장으로 설명합니다.


```python
week_means = login_matrix.mean(axis=0)
member_means = login_matrix.mean(axis=1)
print("주별 평균:", week_means)
print("회원별 평균:", member_means)
assert np.allclose(member_means, [2.0,0.5,2.5,0.5])
```

## Q5. 조건 mask와 np.where로 활동 상태를 만드세요

> **처음 배우는 사람을 위한 안내**
>
> 뜻: 비교 결과인 True·False를 선택 도구로 쓰고, np.where는 조건에 따라 두 값 중 하나를 놓습니다.
> 
> 이유: 수백 명을 같은 기준으로 빠르게 분류하기 위해서입니다.
> 
> 정상: 총합 [8,2,10,2], 라벨 [관찰,낮음,우선,낮음]입니다.
> 
> 복구: 회원별 합계는 axis=1이고 np.where의 순서는 조건, 참일 때 값, 거짓일 때 값입니다.

### 실행 전에 생각할 질문

우선 기준을 10으로 예상한 후 8로 바꿔 몇 명이 바뀌는지 실험하세요.

### 학생이 할 일

1. 결과를 먼저 적습니다.
2. 핵심 빈칸만 채웁니다.
3. 실행 결과를 예상과 비교합니다.
4. 숫자·조건·열 하나를 바꿔 다시 실행합니다.
5. 결과가 달라진 이유를 한 문장으로 설명합니다.


```python
totals = login_matrix.sum(axis=1)
priority_mask = totals >= 10
labels = np.where(priority_mask, "우선", np.where(totals >= 5, "관찰", "낮음"))
print(totals)
print(labels)
assert labels.tolist() == ["관찰","낮음","우선","낮음"]

labels_at_8 = np.where(totals >= 8, "우선", np.where(totals >= 5, "관찰", "낮음"))
print("기준 8:", labels_at_8)
```

## Q6. reshape와 broadcasting 결과를 먼저 예상하세요

> **처음 배우는 사람을 위한 안내**
>
> 뜻: reshape는 값 수를 유지한 채 모양을 바꾸고, broadcasting은 작은 값 묶음을 각 행에 맞춰 계산합니다.
> 
> 이유: 데이터가 모델이 요구하는 모양인지 맞추고 주차별 보정값을 한 번에 적용하기 위해서입니다.
> 
> 정상: np.arange(8).reshape(2,4)는 2×4, 각 행에 [0,1,0,1]이 더해집니다.
> 
> 복구: reshape 전후 전체 값 개수가 같은지 확인합니다.

### 실행 전에 생각할 질문

이 부분은 강사 시연입니다. 학생은 코드 작성보다 결과 모양을 말합니다.

### 학생이 할 일

1. 결과를 먼저 적습니다.
2. 핵심 빈칸만 채웁니다.
3. 실행 결과를 예상과 비교합니다.
4. 숫자·조건·열 하나를 바꿔 다시 실행합니다.
5. 결과가 달라진 이유를 한 문장으로 설명합니다.


```python
numbers = np.arange(8).reshape(2, 4)
bonus_by_week = np.array([0,1,0,1])
print(numbers)
print(numbers + bonus_by_week)
assert numbers.shape == (2,4)
```

## Q7. 도움 없이 새 문제 풀기 — 네 회원의 최근 2주 합계와 우선 여부

> **처음 배우는 사람을 위한 안내**
>
> 성공 조건: 최근 2주를 슬라이싱하고, 회원별 합계를 구하고, 5회 이상을 ‘최근활발’로 표시합니다.
> 
> 힌트 정책: 1차에는 힌트 없음 → 막히면 ‘열 2:4’ → 다시 막히면 ‘axis=1’만 제공합니다.
> 
> 정상: 박민수만 최근활발입니다.
> 
> 복구: 중간 결과의 shape를 한 줄씩 출력해 처음 달라지는 지점을 찾습니다.

### 실행 전에 생각할 질문

앞 셀을 복사하지 말고 선택→합계→분류 세 줄을 직접 작성하세요.

### 학생이 할 일

1. 결과를 먼저 적습니다.
2. 핵심 빈칸만 채웁니다.
3. 실행 결과를 예상과 비교합니다.
4. 숫자·조건·열 하나를 바꿔 다시 실행합니다.
5. 결과가 달라진 이유를 한 문장으로 설명합니다.


```python
last_two = login_matrix[:, 2:4]
last_two_totals = last_two.sum(axis=1)
recent_labels = np.where(last_two_totals >= 5, "최근활발", "일반")
for name, total, label in zip(member_names, last_two_totals, recent_labels):
    print(name, total, label)
assert recent_labels.tolist() == ["일반","일반","최근활발","일반"]
```
