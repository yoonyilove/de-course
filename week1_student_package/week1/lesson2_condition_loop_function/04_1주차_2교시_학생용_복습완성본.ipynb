## 0. 오늘 사용할 연습 데이터 만들기 · 정상 출력

정상: `준비된 회원 수: 4`, 첫 회원 김철수와 문자열 리스트가 보입니다.

복구: NameError가 나면 이 셀을 먼저 실행했는지 확인합니다.

> **데이터 구분:** `weekly_logins_raw`는 실제 CSV의 열이 아니라 이 교시에서 만든 dict의 key입니다. 실제 CSV에는 `login_week1`부터 `login_week4`까지 네 개의 숫자 열이 있습니다. 여기서는 외부 시스템이 네 값을 문자열로 전달한 상황을 가정하여 `int()` 변환을 연습합니다.


```python
members = [
    {
        "member_id": "M001", "name": "김철수", "age": 38,
        "is_profile_active": True,
        "weekly_logins_raw": ["2", "3", "2", "1"]
    },
    {
        "member_id": "M002", "name": "이영희", "age": 34,
        "is_profile_active": False,
        "weekly_logins_raw": ["1", "0", "1", "0"]
    },
    {
        "member_id": "M003", "name": "박민수", "age": 36,
        "is_profile_active": True,
        "weekly_logins_raw": ["1", "2", "4", "3"]
    },
    {
        "member_id": "M004", "name": "최하늘", "age": 31,
        "is_profile_active": False,
        "weekly_logins_raw": ["1", "0", "0", "1"]
    },
]

print("준비된 회원 수:", len(members))
print("첫 회원:", members[0]["name"], members[0]["weekly_logins_raw"])
```

## Q1 정답 · int와 str

말할 문장: “따옴표가 붙으면 모양이 숫자여도 글자입니다. 계산 직전에 `int`, 문장에 붙일 때 `str`을 씁니다.”

> **`str()`이 꼭 필요한 때:** 숫자를 사람이 본다는 이유만으로 `str()`로 바꿀 필요는 없습니다. `print("합계:", total)`이나 `f"합계: {total}"`은 숫자를 그대로 넣어도 됩니다. `"합계: " + total`처럼 문자열과 숫자를 `+`로 직접 이어 붙일 때만 `str(total)`이 필요합니다.

```python
prediction_1 = "23"
prediction_2 = "5"
print("내 예상:", prediction_1, "/", prediction_2)
print("실제 1:", "2" + "3")
print("실제 2:", int("2") + int("3"))
```

```python
member = members[0]
raw_counts = member["weekly_logins_raw"]
weekly_total = int(raw_counts[0]) + int(raw_counts[1]) + int(raw_counts[2]) + int(raw_counts[3])
message = member["name"] + " 4주 로그인 합계: " + str(weekly_total) + "회"
print(message)
```

## Q3 정답 · def / return

뜻: `def`는 규칙에 이름을 붙이고, `return`은 함수 밖에서 다시 쓸 결과를 돌려줍니다.

복구: 함수 코드를 고친 뒤에는 정의 셀을 다시 실행해야 새 규칙이 커널에 반영됩니다.

```python
def activity_level(login_counts, high_boundary=10, medium_boundary=5):
    total = sum(int(x) for x in login_counts)
    if total >= high_boundary:
        level = "활발"
    elif total >= medium_boundary:
        level = "보통"
    else:
        level = "낮음"
    return total, level

total, level = activity_level(members[0]["weekly_logins_raw"])
print(total, level)
```

```python
def fixed_activity_level(total):
    if total >= 10:
        level = "활발"
    elif total >= 5:
        level = "보통"
    else:
        level = "낮음"
    return level

result = fixed_activity_level(8)
print("함수 밖의 결과:", result)
```

## Q4 정답 · for

정상: 네 줄이 나오며 박민수만 활발입니다.

학생이 `range(len(members))`를 먼저 쓰면 틀렸다고 하지 말고, “지금 위치 번호가 꼭 필요한가?”를 질문해 `for member in members`가 더 직접적임을 비교합니다.

```python
predicted_rows = 4
predicted_active_name = "박민수"
reports = []
for member in members:
    total, level = activity_level(member["weekly_logins_raw"])
    reports.append({
        "member_id": member["member_id"],
        "name": member["name"],
        "login_total": total,
        "activity_level": level,
    })
print("예상 행 수:", predicted_rows, "실제 행 수:", len(reports))
for report in reports:
    print(report)
```

```python
members_changed = [member.copy() for member in members]
members_changed[3]["weekly_logins_raw"] = ["5", "5", "0", "0"]
for member in members_changed:
    total, level = activity_level(member["weekly_logins_raw"])
    print(member["name"], total, level)
```

## Q5 정답 · range

뜻: `range`는 반복 그 자체가 아니라 필요한 정수의 흐름을 만듭니다. 위치가 필요 없는 반복에는 굳이 쓰지 않습니다.

```python
for index in range(len(members)):
    member = members[index]
    total, level = activity_level(member["weekly_logins_raw"])
    display_number = index + 1
    print(str(display_number) + "번 " + member["name"] + "님: " + level)
```

```python
predicted_indexes = [1, 3]
actual_indexes = list(range(1, len(members), 2))
print("예상 위치:", predicted_indexes)
print("실제 위치:", actual_indexes)
print("선택 회원:", [members[i]["name"] for i in actual_indexes])
```

## Q6 오류 원인 찾기 · 수정본

```python
name = "김철수"
total = 8
print(name + "님의 로그인 합계는 " + str(total) + "회입니다.")

print(activity_level(["2", "3", "2", "1"]))

def no_return_level(total):
    if total >= 10:
        level = "활발"
    elif total >= 5:
        level = "보통"
    else:
        level = "낮음"
    return level

saved_level = no_return_level(8)
print(saved_level)
```

## Q7 도움 없이 새 문제 풀기 · 한 가지 정답

다른 올바른 풀이도 허용합니다. 평가 기준은 함수 입력 사용, 정수 변환, 조건 두 개, `return`, 네 회원 반복입니다.

```python
def consultation_priority(member):
    total = sum(int(x) for x in member["weekly_logins_raw"])
    if member["is_profile_active"] and total >= 5:
        return "상담우선"
    return "일반"

for member in members:
    print(member["name"], consultation_priority(member))
```
