# 2주차 1교시 · 이름 규칙으로 필요한 CSV 25개 찾기

**학생용 복습완성본**

오늘의 질문: **CSV 25개의 이름을 하나씩 쓰지 않고 필요한 파일만 정확히 찾으려면 어떻게 해야 할까요?**

오늘의 완성 결과: 01번부터 25번까지 정렬된 CSV 위치 25개를 batch_files에 저장합니다.

학생이 막히면 정답 전체를 바로 보여주지 말고, [자료구조] → [입력] → [한 줄씩 읽기] → [정상 결과] 순으로 힌트를 줍니다.


```python
from pathlib import Path

def find_week2_data_dir():
    """노트북을 어느 교시 폴더에서 열어도 2주차 데이터 폴더를 찾습니다."""
    here = Path.cwd().resolve()
    for base in [here, *here.parents]:
        candidates = [
            base / "course_materials" / "matchmaking_data" / "02_week2",
            base / "matchmaking_data" / "02_week2",
            base / "02_week2",
        ]
        for candidate in candidates:
            if (candidate / "activity_batches").exists():
                return candidate
    raise FileNotFoundError("02_week2 데이터 폴더를 찾지 못했습니다. python_lecture 프로젝트 안에서 노트북을 열어 주세요.")

DATA_DIR = find_week2_data_dir()
print("데이터 폴더:", DATA_DIR)
```

## Q1. 파일을 찾기 전에 데이터 폴더가 실제로 있는지 확인합니다

**먼저 예상하기**

- CSV를 찾기 전에 폴더 위치부터 확인하는 이유는 무엇일까요?

**왜 이 일을 하나요?** 경로가 틀린 문제를 CSV 내용 문제와 구분합니다.

**현재 자료구조** DATA_DIR과 batch_dir은 파일 내용이 아니라 위치를 나타내는 Path 객체입니다.

**코드 읽기 도움** Path는 주소를 글자로만 다루지 않고, 주소를 잇고 확인하는 기능을 함께 가진 객체입니다.

**정상 결과** activity_batches로 끝나는 경로와 True

**오류가 나면** False. DATA_DIR과 실제 폴더명을 한 글자씩 비교합니다.


```python
batch_dir = DATA_DIR / "activity_batches"
print(batch_dir)
print(batch_dir.exists())
```

### 강사 설명 메모

- 기대 답: 파일을 열기 전에 폴더 경로가 맞는지 확인하면 경로 오류와 CSV 내용 오류를 구분할 수 있습니다.
- 그대로 말할 문장: “Path는 주소를 글자로만 다루지 않고, 주소를 잇고 확인하는 기능을 함께 가진 객체입니다.”
- 결과 확인: activity_batches로 끝나는 경로와 True
- 복구 순서: DATA_DIR과 실제 폴더명을 한 글자씩 비교합니다.


## Q2. 같은 이름 규칙을 가진 CSV를 한꺼번에 찾습니다

**먼저 예상하기**

- 01, 02처럼 파일마다 달라지는 부분은 어떻게 표시할까요?
- 내 예상: ____________________

**왜 이 일을 하나요?** 같은 규칙으로 이름 붙은 여러 파일을 한 번에 찾습니다.

**현재 자료구조** found는 Path 객체들이 차례로 들어 있는 list입니다.

**코드 읽기 도움** glob은 폴더 안에서 이름 규칙에 맞는 경로를 찾습니다. 별표는 01, 02 같은 달라지는 부분을 대신합니다.

**정상 결과** <class 'list'> 25

- 학생에게 물을 질문: 01, 02처럼 파일마다 달라지는 부분은 어떻게 표시할까요?
- 기대 답: 파일마다 달라지는 부분을 별표(*)로 표시합니다.
- 그대로 말할 문장: “glob은 폴더 안에서 이름 규칙에 맞는 경로를 찾습니다. 별표는 01, 02 같은 달라지는 부분을 대신합니다.”
- 결과 확인: <class 'list'> 25
- 복구 순서: 패턴의 앞부분·확장자와 실제 파일명을 비교합니다.

## Q4. 01번부터 25번까지 빠진 파일이 없는지 이름으로 확인합니다

**먼저 예상하기**

- 파일이 25개라면 01번부터 25번까지 모두 있다고 확신할 수 있을까요?
- 내 예상: ____________________

**왜 이 일을 하나요?** 개수만 25인 우연을 피하고 01~25가 빠짐없이 있는지 확인합니다.

**현재 자료구조** expected_names와 actual_names는 문자열 set입니다. missing은 기대했지만 실제로 없는 이름 set입니다.

**코드 읽기 도움** 집합의 빼기는 왼쪽에는 있지만 오른쪽에는 없는 항목을 찾습니다.

**정상 결과** 25 set()

**오류가 나면** missing에 파일명이 표시됨. 표시된 파일이 폴더에 있는지, 이름이 정확한지 확인합니다.


```python
expected_names = {f"activity_batch_{i:02d}.csv" for i in range(1, 26)}
actual_names = {p.name for p in batch_files}
missing = expected_names - actual_names
print(len(actual_names), missing)
```

### 강사 설명 메모

- 학생에게 물을 질문: 파일이 25개라면 01번부터 25번까지 모두 있다고 확신할 수 있을까요?
- 기대 답: 아니요. 한 번호가 빠지고 다른 파일이 하나 더 있어도 전체 개수는 25개일 수 있습니다.
- 그대로 말할 문장: “집합의 빼기는 왼쪽에는 있지만 오른쪽에는 없는 항목을 찾습니다.”
- 결과 확인: 25 set()
- 복구 순서: 표시된 파일이 폴더에 있는지, 이름이 정확한지 확인합니다.


## Q5. 잘못된 파일명 규칙을 고쳐 CSV 25개를 다시 찾습니다

**먼저 예상하기**

- 검색 결과가 0개라면 가장 먼저 무엇을 확인해야 할까요?
- 내 예상: ____________________

**왜 이 일을 하나요?** glob 결과가 0개이거나 너무 많을 때 패턴을 스스로 조정합니다.

**현재 자료구조** wrong과 fixed는 각각 Path list이며 원본 파일은 바뀌지 않습니다.

**코드 읽기 도움** assert는 결과가 반드시 25여야 한다는 안전벨트입니다.

- 복구 순서: len(fixed)를 먼저 출력하고 패턴·폴더·파일명을 확인합니다.
